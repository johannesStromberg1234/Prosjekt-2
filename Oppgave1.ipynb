Hvis temperaturen senkes slik at $\beta k = 1000$, betyr det at 
$\beta = \frac{1}{k_B T}$ blir ganske stor. Det tilsvarer en situasjon
med lav temperatur. Sannsynlighetsfordelingen er gitt av
Boltzmann-faktoren

$$
P(x) \propto e^{-\beta U(x)}.
$$

I oppgave 2a og 2b er det potensialet $U(x)$ som bestemmer formen på
fordelingen. Når $\beta$ blir stor, vil uttrykket $e^{-\beta U(x)}$
gjøre at tilstander med energi høyere enn minimum blir mye mindre
sannsynlige. Derfor vil fordelingen i praksis samle seg rundt punktet
der potensialet er minst.

For å forstå dette litt bedre kan vi se på potensialet nær dette
punktet. Hvis vi kaller minimumet $x_0$, kan vi utvikle potensialet
rundt dette punktet ved en Taylorutvikling:

$$
U(x) \approx U(x_0) + \frac12 U''(x_0)(x-x_0)^2.
$$

Setter vi dette inn i Boltzmann-faktoren får vi

$$
P(x) \propto e^{-\beta U(x_0)}
e^{-\frac12 \beta U''(x_0)(x-x_0)^2}.
$$

Dette uttrykket har samme form som en gaussisk fordeling

$$
P(x) \propto e^{-\frac{(x-\mu)^2}{2\sigma^2}}.
$$

Ved å sammenligne koeffisienten foran kvadratet i eksponenten ser vi at

$$
\frac{1}{2\sigma^2} = \frac12 \beta U''(x_0).
$$

Dette gir

$$
\frac{1}{\sigma^2} = \beta U''(x_0),
$$

og dermed

$$
\sigma = \frac{1}{\sqrt{\beta U''(x_0)}}.
$$

Når $\beta k = 1000$, blir denne bredden veldig liten. Det betyr at
fordelingen er veldig smal rundt $x_0$, og systemet vil nesten alltid
ligge nær dette punktet. Avvik fra minimumet blir derfor lite
sannsynlige.

Fysisk betyr dette at systemet som regel vil ligge nær den mest stabile
tilstanden, altså der potensialenergien er lavest. Når systemet beveger
seg bort fra dette punktet, øker energien, og ved lav temperatur blir
slike tilstander derfor lite sannsynlige. Systemet vil derfor stort sett
holde seg nær minimumet av potensialet og bare gjøre små svingninger
rundt dette punktet på grunn av termiske fluktuasjoner.
